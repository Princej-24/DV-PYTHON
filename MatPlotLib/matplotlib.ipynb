```python
import matplotlib.pyplot as plt

months=["Jan","Feb","Mar","Apr","May"]
sales=[100,150,120,180,200]

plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
#Change Line System
plt.plot(
    months,
    sales,
    color="red",
    linestyle="--",
    linewidth=2,
    marker="o"
)
```

```python
months=["Jan","Feb","Mar","Apr","May"]
sales_2025=[100,150,120,180,200]
sales_2026=[122,170,250,200,250]
plt.plot(months,sales_2025,marker="o")
plt.plot(months,sales_2026,marker="s")
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend("center")
plt.show()
```

```python
import matplotlib.pyplot as plt
students=["Vikky","Rahul","Prateek","Priyanshi","Muskan","Payal"]
marks=[82,45,58,58,98,98]

plt.bar(students,marks,width=0.6)
plt.title("Student Marks")
plt.xlabel("Student")
plt.ylabel("Marks")
plt.show()
```

```python
#Horizontal Bar Chart
plt.barh(students,marks)
plt.show()
```

```python
#Multiple Bar Charts
students=["A","B","C","D"]
python=[80,20,64,84]
java=[80,20,64,84]
import numpy as np
x=np.arange(len(students))
width=0.35

plt.bar(x-width/2,python,width,label="Python")
plt.bar(x+width/2,java,width,label="Java")
plt.xticks(x,students)
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
subjects = ["AI", "Python", "Java", "SQL"]

hours = [35,25,20,20]

explode = [0.1, 0, 0, 0]
mycolors=["Red","LightPink","Blue","Orange"]

plt.pie(hours,

labels=subjects,

autopct="%1.1f%%", #Percentage in Pie Chart

explode=explode, #Explode Pie Chart

shadow = True, #add shaddw

colors = mycolors #add color
)
plt.title("Study time")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
height =[150,155,160,165,170]
weight =[50,55,60,65,70]

plt.scatter(height,weight,
 s=100,#Scatter Plot Size
 alpha=0.5 #Scatter Plot Transparency
 )
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height Vs Weight")
plt.show()
```

```python
#Histogram
#show how data is distributed

import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90, 95]

plt.hist(marks, bins=5)

plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Distribution of Marks")

plt.show()
```

```python
#BoxPlot
#outlier detection

import matplotlib.pyplot as plt

marks = [45, 50, 55, 60, 62, 65, 70, 72, 75, 80, 85, 90]

plt.boxplot(marks)

plt.title("Box Plot of Marks")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

python = [70, 75, 80, 85, 90]
java = [60, 65, 70, 75, 80]
cpp = [65, 70, 75, 80, 85]

plt.boxplot(
    [python, java, cpp],
    tick_labels=["Python", "Java", "C++"]
)

plt.show()
```

```python
#Adds grid lines to improve readability 
import numpy as np
import matplotlib.pyplot as plt

x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

plt.figure(figsize=(5,2))
plt.title("Sports Watch Data")
plt.xlabel("Average Pulse")
plt.ylabel("Calorie Burnage")

plt.plot(x, y)

plt.grid(axis='y')
plt.grid(color='green', linestyle='--', linewidth=0.5)

plt.show()
```
