```python
sns.countplot(
    data=tips,
    x="smoker",
    hue="time"
)
plt.show()
```

```python
#histogram
sns.histplot(
    data=tips,
    x="total_bill"
)
plt.show()
```

```python
#histogram
sns.histplot(
    data=titanic,
    x="age"
)
plt.show()
```

```python
#histogram
sns.histplot(
    data=titanic,
    x="age",
    bins=10,
    kde=True
)
plt.show()
```

```python
#KDE means Kernel Density Estimation,
#it shows probability distribution of a variable
sns.kdeplot(
data=titanic,
x="age",
)
plt.show()
```

```python
sns.boxplot(
    data=titanic,
    x="sex",
    y="age",
    hue="survived"
)
plt.show()
```
